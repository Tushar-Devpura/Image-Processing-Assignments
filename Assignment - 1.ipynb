### 1. Write a Python program to read an image named 'flower.jpg' using OpenCV and display it in a window titled 'Original Image'.


```python
import cv2

img = cv2.imread("flower.jpg")
if img is not None:
    cv2.imshow("Original Image", img)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
else:
    print("Error: Image not found.")
```

### 5. Write a Python program using OpenCV that resizes an image to 50% of its original width and height.


```python
import cv2

img = cv2.imread("image.jpg") # Provide a valid image path
if img is not None:
    height, width = img.shape[:2]
    # Resize to 50%
    resized_img = cv2.resize(img, (width // 2, height // 2))
    print(f"Original shape: {img.shape}")
    print(f"Resized shape: {resized_img.shape}")
else:
    print("Error: Image not found.")
```

### 6. Write a Python code snippet to crop the central 200x200 pixel area of an image and display it using Matplotlib.


```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("image.jpg") # Provide a valid image path
if img is not None:
    # Convert BGR to RGB for correct color display in Matplotlib
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    h, w = img_rgb.shape[:2]
    
    # Calculate starting coordinates for the crop
    start_y = max(0, (h // 2) - 100)
    start_x = max(0, (w // 2) - 100)
    
    cropped = img_rgb[start_y:start_y+200, start_x:start_x+200]
    
    plt.imshow(cropped)
    plt.title("Central 200x200 Crop")
    plt.axis("off")
    plt.show()
else:
    print("Error: Image not found.")
```

### 8. Write a Python statement to flip the image 'car.jpg' left-to-right using OpenCV and display the result.


```python
import cv2

img = cv2.imread("car.jpg")
if img is not None:
    flipped_img = cv2.flip(img, 1) # 1 represents horizontal flipping (left-to-right)
    cv2.imshow("Flipped Image", flipped_img)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
else:
    print("Error: Image not found.")
```

### 9. Write a Python program that converts an image from RGB to HSV colour space, and display each channel (Hue, Saturation, Value) separately using Matplotlib.


```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("image.jpg") # Provide a valid image path
if img is not None:
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img_hsv = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2HSV)
    
    # Split into Hue, Saturation, and Value channels
    h, s, v = cv2.split(img_hsv)

    fig, axs = plt.subplots(1, 3, figsize=(15, 5))
    
    axs[0].imshow(h, cmap="gray")
    axs[0].set_title("Hue")
    axs[0].axis("off")
    
    axs[1].imshow(s, cmap="gray")
    axs[1].set_title("Saturation")
    axs[1].axis("off")
    
    axs[2].imshow(v, cmap="gray")
    axs[2].set_title("Value")
    axs[2].axis("off")
    
    plt.tight_layout()
    plt.show()
else:
    print("Error: Image not found.")
```

### 10. Write a short Python code snippet to compute the Mean Squared Error (MSE) between two grayscale images, 'img1.jpg' and 'img2.jpg'.


```python
import cv2
import numpy as np

img1 = cv2.imread("img1.jpg", cv2.IMREAD_GRAYSCALE)
img2 = cv2.imread("img2.jpg", cv2.IMREAD_GRAYSCALE)

if img1 is not None and img2 is not None:
    if img1.shape == img2.shape:
        mse = np.mean((img1.astype("float") - img2.astype("float")) ** 2)
        print(f"Mean Squared Error (MSE): {mse}")
    else:
        print("Error: Images must have the same dimensions to compute MSE.")
else:
    print("Error: One or both images not found.")
```
